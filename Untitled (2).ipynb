```python
#here categories are get find out
def get_category(description):

    description = description.upper()

    cursor.execute(
        """
        SELECT merchant_keyword, category
        FROM merchant_categories
        """
    )

    merchants = cursor.fetchall()

    for keyword, category in merchants:

        if keyword in description:
            return category

    return None
```

```python
# actual spending per category
expenses = df[df["Type"] == "Debit"]

category_expense = (
    expenses
    .groupby("Category")["Amount"]
    .sum()
    .sort_values(ascending=False)
)

print(category_expense)
```

```python
# create budget datframe
budget_df = category_expense.reset_index()

budget_df.columns = [
    "Category",
    "Spent"
]

budget_df
```

```python
user_budget = {
    "Food": 25000,
    "Shopping": 18000,
    "Transport": 12000,
    "Subscription": 6000,
    "Utilities": 20000,
    "Healthcare": 3000,
    "Entertainment": 4000,
    "Fitness": 4000,
    "Education": 5000
}
```

```python
budget_df["Budget"] = budget_df["Category"].map(user_budget)
```

```python
budget_df["Budget"] = (
    budget_df["Category"]
    .map(user_budget)
)

budget_df["Difference"] = (
    budget_df["Spent"]
    - budget_df["Budget"]
)

budget_df["Status"] = (
    budget_df["Difference"]
    .apply(
        lambda x:
        "Over Budget"
        if x > 0
        else "Within Budget"
    )
)

budget_df
```
